```python
import requests

import cv2
from PIL import Image
import numpy as np


url = "https://raw.githubusercontent.com/jigsawpieces/dog-api-images/main/eskimo/n02109961_10021.jpg"
response = requests.get(url, stream=True)

with open("data/eskimo_dog.jpg", "wb") as f:
    f.write(response.content)

img_pil = Image.open("data/eskimo_dog.jpg")
img_cv2 = cv2.imread("data/eskimo_dog.jpg")
```

```python
img_cv2 = cv2.cvtColor(img_cv2, cv2.COLOR_BGR2RGB) # BGR -> RGB 
```

```python
# не выведет ничего при ~равенстве массивов, иначе - ошибка
np.testing.assert_allclose(img_cv2, img_pil, atol=1e-8)  
# ошибок нет 
```

```python
shape = (224, 224)
img_pil = img_pil.resize(shape)
img_cv2 = cv2.resize(img_cv2, shape)

np.testing.assert_allclose(img_cv2, img_pil, atol=1e-8)
```

```python
img_pil = np.array(img_pil) / 255
img_cv2 = img_cv2 / 255

print(np.min(img_cv2), np.min(img_pil), np.max(img_cv2), np.max(img_pil))
```

```python
import cv2
import requests


mean = [0.4, 0.42, 0.44]
std = [0.2, 0.21, 0.22]

url = "https://raw.githubusercontent.com/jigsawpieces/dog-api-images/main/eskimo/n02109961_10021.jpg"
response = requests.get(url, stream=True)
with open("eskimo_dog.jpg", "wb") as f:
    f.write(response.content)
```

```python
image = cv2.imread("data/eskimo_dog.jpg")
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB) 
image = cv2.resize(image, (356, 356))
image = image / 255
image = (image - mean) / std
np.round(image[0][0][1], 3)
```

```python
print("Размеры до: ", image.shape)

# перемещаем каналы на первую позицию
image = image.transpose(2, 0, 1)
print("Размеры после: ",image.shape) 
```

Оцените качество предобученной модели `maxvit_large_tf_384` из библиотеки timm. На примере одного, уже сохранённого ранее изображения проинициализируйте модель и подготовьте картинку к подаче на вход модели: 
- Загрузите.
- Масштабируйте в нужный формат.
- Нормализуйте сначала в диапазон от 0 до 1, а затем в формат, использованный при тренировке конкретной модели.
- Выведите результирующую размерность изображения.

```python
model = timm.create_model(model_name="maxvit_large_tf_384", pretrained=True)
model.pretrained_cfg
```

```python
def preprocess_image(image, model):
    width, height = model.pretrained_cfg["input_size"][1:3]
    mean = model.pretrained_cfg["mean"]
    std = model.pretrained_cfg["std"]
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB) 
    image = cv2.resize(image, (width, height))
    image = image / 255
    image = (image - mean) / std
    image = image.transpose(2, 0, 1)
    return image

image = cv2.imread("data/eskimo_dog.jpg")
preprocess_image(image, model).shape
```
