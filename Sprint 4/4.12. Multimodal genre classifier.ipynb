# Эксперименты. Сравнение разных методов обучения и оценка важности признаков
В этом уроке мы проведём несколько экспериментов по обучению мультимодальных моделей и сравним их результаты. Для экспериментов возьмём задачу — определить жанр фильма по постеру и текстовому описанию. 

Базовый код обучения модели и сам датасет скачайте по ссылке и расположите в рабочей директории. Архив содержит:
- путь до тренировочной и валидационной выборок `data/train|val.csv`,
- путь до директории с картинками `data/imdb_images`,
- код загрузки и обучения в `dataset.py` и `utils.py` соответственно. Все параметры и пути в коде уже настроены.

Обучайте модели на ВМ с GPU, доступ к которой вы получили в прошлом уроке.
