Напишите функцию `get_sentence_embedding`, которая принимает на вход набор токенов и инициализированную модель `w2v_model`, получает эмбеддинги токенов и возвращает их среднее значение. Сравните полученную косинусную близость двух пар предложений (код для этого уже написан).

```python
import numpy as np
from scipy.spatial.distance import cosine
from gensim.models import KeyedVectors

word2vec_path = "models/word2vec-google-news-300.gz"
w2v_model = KeyedVectors.load_word2vec_format(word2vec_path, binary=True)
```

```python
def get_sentence_embedding(tokens, w2v_model):
    embeds = w2v_model[tokens]
    return np.mean(embeds, axis=0)

  
tokens_A = ["data", "scientists", "are", "cool"]
tokens_B = ["data", "science", "is", "awesome"]
tokens_C = ["clouds", "in", "the", "sky"]

emb_A = get_sentence_embedding(tokens_A, w2v_model)
emb_B = get_sentence_embedding(tokens_B, w2v_model)
emb_C = get_sentence_embedding(tokens_C, w2v_model)

1 - cosine(emb_A, emb_B), 1 - cosine(emb_A, emb_C)
```
