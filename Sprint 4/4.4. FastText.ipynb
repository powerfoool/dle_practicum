Ваша задача — инициализировать FastText модель, скачанную по ссылке выше, и получить вектор-эмбеддинг для text. В конце выведите размерность полученного вектора vec.

```python
from gensim.models import FastText

fasttext_path = 'models/cc.ru.300.bin'
model = FastText.load_fasttext_format(fasttext_path)
```

```python
text = "FastText - это 'быстрый текст'"

vec = model.wv.get_sentence_vector(text.split(" "))
vec.shape
```
